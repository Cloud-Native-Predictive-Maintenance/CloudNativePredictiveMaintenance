```python
unit_1 = df[df["unit_id"] == 1]

plt.figure(figsize=(12, 5))

plt.plot(
    unit_1["cycle"],
    unit_1["sensor_2"]
)

plt.xlabel("Cycle")
plt.ylabel("Sensor 2")
plt.title("Machine 1 - Sensor 2 over Time")
plt.grid(True)

plt.show()
```

```python
sensor_cols = [f"sensor_{i}" for i in range(1, 22)]

sensor_std = df[sensor_cols].std().sort_values(ascending=False)

sensor_std
```

```python
sensor_std.plot(
    kind="bar",
    figsize=(12, 5)
)

plt.title("Sensor Variability")
plt.xlabel("Sensor")
plt.ylabel("Standard Deviation")
plt.xticks(rotation=45)

plt.show()
```

```python
df["RUL"] = (
    df.groupby("unit_id")["cycle"]
    .transform("max")
    - df["cycle"]
)

df[["unit_id", "cycle", "RUL"]].head(10)
```

```python
df.groupby("unit_id")["RUL"].agg(["min", "max"])
```

```python
df["RUL"].describe()
```

```python
sensor_rul_corr = (
    df[sensor_cols + ["RUL"]]
    .corr()["RUL"]
    .drop("RUL")
    .sort_values()
)

sensor_rul_corr
```

```python
top_sensor = sensor_rul_corr.index[0]

print("Most negatively correlated sensor:", top_sensor)
print("Correlation:", sensor_rul_corr.iloc[0])
```

```python
plt.figure(figsize=(10, 5))

plt.scatter(
    df[top_sensor],
    df["RUL"],
    alpha=0.2
)

plt.xlabel(top_sensor)
plt.ylabel("RUL")
plt.title(f"{top_sensor} vs RUL")
plt.grid(True)

plt.show()
```

```python
sensor_rul_corr
```

```python
print("Most negative correlations:")
print(sensor_rul_corr.head(10))

print("\nMost positive correlations:")
print(sensor_rul_corr.tail(10))
```

```python
constant_sensors = [
    sensor for sensor in sensor_cols
    if df[sensor].nunique() <= 1
]

print("Constant sensors:")
print(constant_sensors)
```

```python
useful_sensors = [
    sensor for sensor in sensor_cols
    if sensor not in constant_sensors
]

print("Sensors with variation:")
print(useful_sensors)
print("Number of usable sensors:", len(useful_sensors))
```

```python
model_df = df[["unit_id", "cycle"] + useful_sensors + ["RUL"]].copy()

model_df.head()
```
